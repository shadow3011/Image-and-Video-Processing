```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('/content/bg.jpg', 0)

k = int(input("Enter filter size (odd): "))

# Padding method
pad_type = input("Padding (zero/replicate): ")

if pad_type.lower() == "zero":
    border = cv2.BORDER_CONSTANT
else:
    border = cv2.BORDER_REPLICATE

padded = cv2.copyMakeBorder(
    img, k//2, k//2, k//2, k//2,
    border
)

avg_img = cv2.blur(padded, (k, k))

plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(122)
plt.imshow(avg_img, cmap='gray')
plt.title("Averaging Filter")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('/content/bg.jpg', 0)

kernel = np.array([
    [1,1,1],
    [1,0,1],
    [1,1,1]
], dtype=np.float32)

kernel = kernel / kernel.sum()

weighted_img = cv2.filter2D(
    img,
    -1,
    kernel
)

plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(122)
plt.imshow(weighted_img, cmap='gray')
plt.title("Weighted Filter")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('/content/bg.jpg', 0)

k = int(input("Enter kernel size (odd): "))
sigma = float(input("Enter sigma value: "))

gaussian_img = cv2.GaussianBlur(
    img,
    (k, k),
    sigma
)

plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Original")

plt.subplot(122)
plt.imshow(gaussian_img, cmap='gray')
plt.title("Gaussian Filter")

plt.show()
```
