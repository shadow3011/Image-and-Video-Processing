```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read color image
img = cv2.imread('/content/bg.jpg')

# Convert BGR to RGB (OpenCV reads as BGR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Split channels
R = img_rgb[:, :, 0]
G = img_rgb[:, :, 1]
B = img_rgb[:, :, 2]

# Mean average grayscale
gray = ((R.astype(np.float32) +
         G.astype(np.float32) +
         B.astype(np.float32)) / 3).astype(np.uint8)

# Display
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img_rgb)
plt.title("Original RGB Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale (Mean Average)")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('/content/bg.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Input weights
wr = float(input("Enter weight of R: "))
wg = float(input("Enter weight of G: "))
wb = float(input("Enter weight of B: "))

# Check sum of weights
if abs((wr + wg + wb) - 1) > 0.001:
    print("Error: Sum of weights must be 1")
else:
    R = img_rgb[:, :, 0]
    G = img_rgb[:, :, 1]
    B = img_rgb[:, :, 2]

    gray = (wr * R + wg * G + wb * B).astype(np.uint8)

    plt.figure(figsize=(10,5))

    plt.subplot(1,2,1)
    plt.imshow(img_rgb)
    plt.title("Original RGB Image")
    plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(gray, cmap='gray')
    plt.title("Weighted Grayscale")
    plt.axis('off')

    plt.show()
```
