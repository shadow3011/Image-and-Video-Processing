 Write a program to generate following noises using equations derived
from PDFs of noise distributions. Compare your output with those
generated using inbuilt function imnoise. Plot the histograms of generated
noise to determine the shape of the distribution.

1. Uniform Noise

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("/content/denoising.png", 0)

if img is None:
    print("Image not found!")
    exit()

# Generate uniform noise using equation
a = -30
b = 30

noise = np.random.uniform(a, b, img.shape)

# Add noise to image
noisy_img = np.clip(img + noise, 0, 255).astype(np.uint8)

# Inbuilt function comparison
builtin_noise = np.random.uniform(a, b, img.shape)
builtin_img = np.clip(img + builtin_noise, 0, 255).astype(np.uint8)

# Display images
plt.figure(figsize=(12, 7))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(noisy_img, cmap='gray')
plt.title("Uniform Noise - Equation")

plt.subplot(2, 2, 3)
plt.hist(noise.ravel(), bins=50)
plt.title("Uniform Noise Histogram")

plt.subplot(2, 2, 4)
plt.hist(builtin_noise.ravel(), bins=50)
plt.title("Inbuilt Uniform Noise Histogram")

plt.tight_layout()
plt.show()
```

2. Gaussian Noise

3. Erlang Noise

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import math

img = cv2.imread("/content/denoising.png", 0)

if img is None:
    print("Image not found!")
    exit()

# Parameters
a = 0.5
b = 2

# Generate Erlang noise using equation
u = np.random.uniform(0, 1, img.shape)

noise = -np.log(
    np.random.uniform(0, 1, img.shape) *
    np.random.uniform(0, 1, img.shape)
) / a

# Inbuilt Gamma function (Erlang when b is integer)
builtin_noise = np.random.gamma(b, 1/a, img.shape)

# Add noise to image
noisy_img = np.clip(img + noise, 0, 255).astype(np.uint8)
builtin_img = np.clip(img + builtin_noise, 0, 255).astype(np.uint8)

# Display
plt.figure(figsize=(12, 7))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(noisy_img, cmap='gray')
plt.title("Erlang Noise - Equation")

plt.subplot(2, 2, 3)
plt.hist(noise.ravel(), bins=50)
plt.title("Erlang Noise Histogram")

plt.subplot(2, 2, 4)
plt.hist(builtin_noise.ravel(), bins=50)
plt.title("Inbuilt Gamma Histogram")

plt.tight_layout()
plt.show()
```

4. Exponential Noise

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/denoising.png", 0)

if img is None:
    print("Image not found!")
    exit()

# Parameter
a = 0.05

# Generate exponential noise using inverse CDF
u = np.random.uniform(0, 1, img.shape)
noise = -np.log(1 - u) / a

# Inbuilt function
builtin_noise = np.random.exponential(1/a, img.shape)

# Add noise
noisy_img = np.clip(img + noise, 0, 255).astype(np.uint8)
builtin_img = np.clip(img + builtin_noise, 0, 255).astype(np.uint8)

# Display
plt.figure(figsize=(12, 7))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(noisy_img, cmap='gray')
plt.title("Exponential Noise - Equation")

plt.subplot(2, 2, 3)
plt.hist(noise.ravel(), bins=50)
plt.title("Exponential Noise Histogram")

plt.subplot(2, 2, 4)
plt.hist(builtin_noise.ravel(), bins=50)
plt.title("Inbuilt Exponential Histogram")

plt.tight_layout()
plt.show()
```

5. Rayleigh Noise

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("/content/denoising.png", 0)

if img is None:
    print("Image not found!")
    exit()

# Parameters
a = 0
b = 100

# Generate Rayleigh noise using inverse CDF
u = np.random.uniform(0, 1, img.shape)

noise = a + np.sqrt(-b * np.log(1-u))

# Inbuilt function
builtin_noise = np.random.rayleigh(np.sqrt(b/2), img.shape)

# Add noise
noisy_img = np.clip(img + noise, 0, 255).astype(np.uint8)
builtin_img = np.clip(img + builtin_noise, 0, 255).astype(np.uint8)

# Display
plt.figure(figsize=(12, 7))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(2, 2, 2)
plt.imshow(noisy_img, cmap='gray')
plt.title("Rayleigh Noise - Equation")

plt.subplot(2, 2, 3)
plt.hist(noise.ravel(), bins=50)
plt.title("Rayleigh Noise Histogram")

plt.subplot(2, 2, 4)
plt.hist(builtin_noise.ravel(), bins=50)
plt.title("Inbuilt Rayleigh Histogram")

plt.tight_layout()
plt.show()
```
