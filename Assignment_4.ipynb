```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread('/content/bg.jpg', 0)

# Log transformation
log_img = np.log1p(img.astype(np.float32))

# Normalize to 0-255
log_img = cv2.normalize(log_img, None, 0, 255, cv2.NORM_MINMAX)
log_img = np.uint8(log_img)

# Display
plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(122)
plt.imshow(log_img, cmap='gray')
plt.title("Log Transformed")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('/content/bg.jpg', 0)

# Input gamma value
gamma = float(input("Enter gamma value: "))

# Normalize image to [0,1]
img_norm = img / 255.0

# Gamma correction
gamma_img = np.power(img_norm, gamma)

# Convert back to [0,255]
gamma_img = np.uint8(gamma_img * 255)

# Display
plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(122)
plt.imshow(gamma_img, cmap='gray')
plt.title(f"Gamma = {gamma}")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread('/content/bg.jpg', 0)

# Input ranges
r1 = int(input("Enter input lower limit (r1): "))
r2 = int(input("Enter input upper limit (r2): "))

s1 = int(input("Enter output lower limit (s1): "))
s2 = int(input("Enter output upper limit (s2): "))

# Contrast stretching
stretched = np.zeros_like(img, dtype=np.float32)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        r = img[i, j]

        if r < r1:
            stretched[i, j] = 0
        elif r > r2:
            stretched[i, j] = 255
        else:
            stretched[i, j] = ((r - r1) * (s2 - s1) / (r2 - r1)) + s1

stretched = np.uint8(stretched)

# Display
plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(122)
plt.imshow(stretched, cmap='gray')
plt.title("Contrast Stretched")
plt.axis('off')

plt.show()
```
